### Joins

### Inner Join 

```python
dataj1 = [('1','gaur','d01'),
          ('2','kit','d02'),
          ('3','sam','d03'),
          ('4','tim','d03'),
          ('5','aman','d05'),
          ('6','nad','d06')] 

schemaj1 = 'emp_id STRING, emp_name STRING, dept_id STRING' 

df1 = spark.createDataFrame(dataj1,schemaj1)

dataj2 = [('d01','HR'),
          ('d02','Marketing'),
          ('d03','Accounts'),
          ('d04','IT'),
          ('d05','Finance')]

schemaj2 = 'dept_id STRING, department STRING'

df2 = spark.createDataFrame(dataj2,schemaj2)
```

```python
df1.display()
df2.display()
```

```python
df1.join(df2,df1['dept_id'] == df2['dept_id'],"inner").display()
```

### Window Functions 

### Row Number

```python
from pyspark.sql.window import Window

```

```python
df.withColumn('Row_col', row_number().over(Window.orderBy(desc('Item_Identifier')))).display()
```

### Rank Vs Dense Rank

```python
df.select("Item_Identifier").withColumn('Rank',rank().over(Window.orderBy(desc(col('Item_Identifier'))))).\
  withColumn('Dense_rank',dense_rank().over(Window.orderBy(desc(col('Item_Identifier'))))).\
  withColumn('Percent_rank',percent_rank().over(Window.orderBy(desc(col('Item_Identifier'))))).\
  withColumn('Cume_dist',cume_dist().over(Window.orderBy(desc(col('Item_Identifier'))))).\
  display()
```

### Cumulative Sum

showing overall sum

```python

df.withColumn('cumsum',sum('Item_MRP').over(Window.orderBy('Item_Type'))).display()
```

### showing cumulative sum

Current row

```python

df.select("Item_Type","Item_MRP").withColumn('cumsum',sum('Item_MRP').over(Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding,Window.currentRow))).display()
```

### Unbounded Following

```python
df.select("Item_Type","Item_MRP").withColumn('Total Sum',sum('Item_MRP').over(Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding,Window.unboundedFollowing))).display()
```
